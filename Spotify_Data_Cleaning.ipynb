**Handelling the missing values for categorical columns**

```python
cat_cols = df.select_dtypes(include=['object', 'category']).columns
cat_cols
```

```python
#  categorical columns have missing values 
miss_cat_columns = [col for col in cat_cols if df[col].isna().sum() > 0]

print('columns have missing values', miss_cat_columns)
```

```python
df['artist_name'] = df['artist_name'].fillna('Unknown Artist')
df['genre'] = df['genre'].fillna('Unknown Genre')
df['language'] = df['language'].fillna('Unknown')
df['album_type'] = df['album_type'].fillna('Unknown')
```

**replace missing values with median**

```python
for col in num_col:

    if df[col].isna().sum() > 0:

        median_val = df[col].median()

        df[col] = df[col].fillna(median_val)
        median_val
```

```python
df.isna().sum()
```

**Handeling the outliers**

```python
for col in num_col:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f'{col}: {len(outliers)} outliers')
```

**impossible value -> NaN -> fill with the median**

```python
# Songs shorter than 30 seconds are most likely errors
df.loc[df['duration_ms'] < 30000, 'duration_ms'] = np.nan
df['duration_ms'] = df['duration_ms'].fillna(df['duration_ms'].median())
```

```python
df['duration_ms'].plot(kind='box')
```

```python
Q1 = df['duration_ms'].quantile(0.25)
Q3 = df['duration_ms'].quantile(0.75)
IQR = Q3 - Q1
lb = Q1 - 1.5 * IQR
ub= Q3 + 1.5 * IQR
df['duration_ms'] = df['duration_ms'].clip(lower=lb, upper=ub)
```

```python
#transform this colums to make understanding easier
df['duration_sec'] = df['duration_ms'] / 1000
```

```python
# Negative values are errors, and 150 billion repeated with the exact same value is a placeholder
df.loc[df['stream_count'] < 0, 'stream_count'] = np.nan
df.loc[df['stream_count'] == 1.5e11, 'stream_count'] = np.nan

df['stream_count'] = df['stream_count'].fillna(
    df['stream_count'].median()
)
print(df['stream_count'].min())
```

```python
# The number of playlists cannot be negative
df.loc[df['in_playlists'] < 0, 'in_playlists'] = np.nan

df['in_playlists'] = df['in_playlists'].fillna(
    df['in_playlists'].median()
)
```

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==========================================
# 0. Rebuild the 6 charts (skip this part if fig1..fig6 already exist)
# ==========================================
fig1 = px.scatter(df.sample(1500, random_state=1), x='tempo', y='energy', color='genre')

g = df.groupby('genre', as_index=False)['stream_count'].mean().sort_values('stream_count', ascending=False)
fig2 = px.bar(g, x='genre', y='stream_count', color_discrete_sequence=['#1DB954'])

fig3 = px.pie(df, names='language')

fig4 = px.box(df, x='language', y='duration_ms')

fig5 = px.histogram(df, x='popularity', color='album_type', nbins=20)

yearly = df['release_year'].value_counts().sort_index().reset_index()
yearly.columns = ['release_year', 'tracks']
fig6 = px.line(yearly, x='release_year', y='tracks')

# ==========================================
# 1. Dashboard layout (3 rows x 4 cols)
# ==========================================
fig = make_subplots(
    rows=3, cols=4,
    specs=[
        [{"type": "domain"}, {"type": "domain"}, {"type": "domain"}, {"type": "domain"}],  # 4 KPIs
        [{"type": "xy", "colspan": 2}, None, {"type": "xy"}, {"type": "xy"}],              # fig1 (wide), fig2, fig6
        [{"type": "xy", "colspan": 2}, None, {"type": "xy"}, {"type": "domain"}],          # fig4 (wide), fig5, fig3
    ],
    subplot_titles=("", "", "", "",
                    "Tempo vs Energy", "Avg Streams per Genre", "Tracks Released per Year",
                    "Duration by Language", "Popularity by Album Type", "Language Share"),
    vertical_spacing=0.12, horizontal_spacing=0.06,
    row_heights=[0.15, 0.42, 0.43],
)

# ==========================================
# 2. The 4 KPIs
# ==========================================
kpis = [
    (len(df),                              "Total Tracks",       {"valueformat": ",.0f"}),
    (df["stream_count"].sum() / 1e9,       "Total Streams",      {"suffix": "B", "valueformat": ",.1f"}),
    (df["popularity"].mean(),              "Average Popularity", {"valueformat": ".1f"}),
    (df["duration_ms"].mean() / 60000,     "Average Duration",   {"suffix": " min", "valueformat": ".2f"}),
]
for col, (value, title, fmt) in enumerate(kpis, start=1):
    fig.add_trace(go.Indicator(mode="number", value=value,
                               title={"text": title}, number=fmt), row=1, col=col)

# ==========================================
# 3. Add the 6 charts
# ==========================================
for t in fig1.data: fig.add_trace(t, row=2, col=1)   # scatter (wide)
for t in fig2.data: fig.add_trace(t, row=2, col=3)   # bar
for t in fig6.data: fig.add_trace(t, row=2, col=4)   # line
for t in fig4.data: fig.add_trace(t, row=3, col=1)   # box (wide)
for t in fig5.data: fig.add_trace(t, row=3, col=3)   # histogram
for t in fig3.data: fig.add_trace(t, row=3, col=4)   # pie

# ==========================================
# 4. Final formatting
# ==========================================
fig.update_traces(textinfo="percent+label", selector=dict(type="pie"))
fig.update_layout(
    template="plotly_dark",
    height=1200,
    title_text="Spotify Analytics Dashboard",
    title_x=0.5,
    showlegend=False,
    barmode="group",
)
fig.update_xaxes(tickangle=45)

fig.show()
```
